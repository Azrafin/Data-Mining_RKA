## **Lihat Kemiringan Distribusi Data**

```python
# 1. Ambil seluruh kolom numerik murni dari data sebelum imputasi
num_cols = df.select_dtypes(include=["int64", "float64"]).columns.tolist()

# Keluarkan 'Id' dan 'SalePrice' jika ada
num_cols = [col for col in num_cols if col not in ["Id", "SalePrice"]]

# 2. Hitung nilai Skewness untuk setiap fitur numerik
skewness_df = (
    pd.DataFrame(
        {
            "Feature": num_cols,
            "Skewness": [df[col].skew() for col in num_cols],
            "Missing_Count": [df[col].isnull().sum() for col in num_cols],
        }
    )
    .sort_values(by="Skewness", ascending=False)
    .reset_index(drop=True)
)

print("=== Rekapitulasi Skewness Seluruh Fitur Numerik ===")
print(skewness_df)

# 3. Kategori Skewness
high_skew = skewness_df[skewness_df["Skewness"].abs() > 1.0]
mod_skew = skewness_df[skewness_df["Skewness"].abs().between(0.5, 1.0)]
sym_skew = skewness_df[skewness_df["Skewness"].abs() < 0.5]

print("\n=== Ringkasan Kategori Skewness ===")
print(f"Fitur Sangat Miring (|Skew| > 1.0) : {len(high_skew)} fitur")
print(f"Fitur Agak Miring (0.5 <= |Skew| <= 1.0) : {len(mod_skew)} fitur")
print(f"Fitur Simetris / Normal (|Skew| < 0.5) : {len(sym_skew)} fitur")

# 4. Visualisasi Distribusi 4 Fitur Numerik Paling Miring
top_4_skewed = high_skew["Feature"].head(4).tolist()

fig, axes = plt.subplots(2, 2, figsize=(12, 8))
axes = axes.flatten()

for i, col in enumerate(top_4_skewed):
    sns.histplot(df[col], kde=True, ax=axes[i], color="purple")
    axes[i].set_title(f"{col} (Skewness: {df[col].skew():.2f})")

plt.suptitle(
    "Distribusi Fitur Numerik dengan Skewness Tinggi",
    fontsize=14,
    fontweight="bold",
)
plt.tight_layout()
plt.show()






```

# **Pembersihan Data (Handling Missing Values)**

```python
# 1. Hitung persentase missing value per kolom
missing_series = df.isnull().mean() * 100
missing_above_50 = missing_series[missing_series > 50].sort_values(
    ascending=False
)

print("=== Kolom dengan Missing Value > 50% ===")
print(missing_above_50)

# 2. Hapus kolom dengan missing value > 50%
cols_to_drop = missing_above_50.index.tolist()
df_clean = df.drop(columns=cols_to_drop).copy()

print(f"\nJumlah kolom sebelum dipotong: {df.shape[1]}")
print(f"Jumlah kolom setelah dipotong: {df_clean.shape[1]}")

# 3. Imputasi sisa missing values
# - Untuk kolom kategorikal (object): isi dengan Modus atau "None"
# - Untuk kolom numerik: isi dengan Median
for col in df_clean.columns:
    if df_clean[col].isnull().sum() > 0:
        if df_clean[col].dtype == "object":
            # Jika kolom kategorikal seperti FireplaceQu/Garage, isi dengan modus
            df_clean[col] = df_clean[col].fillna(df_clean[col].mode()[0])
        else:
            # Jika kolom numerik seperti LotFrontage, isi dengan median
            df_clean[col] = df_clean[col].fillna(df_clean[col].median())

print(
    f"\nSisa total missing values di seluruh dataframe: {df_clean.isnull().sum().sum()}"
)
```

```python
# 1. Visualisasi Boxplot untuk GrLivArea dan SalePrice
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

sns.boxplot(x=df_clean["GrLivArea"], ax=axes[0], color="skyblue")
axes[0].set_title("Boxplot GrLivArea (Deteksi Outlier)")

sns.boxplot(x=df_clean["SalePrice"], ax=axes[1], color="lightgreen")
axes[1].set_title("Boxplot SalePrice (Deteksi Outlier)")

plt.tight_layout()
plt.show()

# 2. Identifikasi outlier ekstrem pada GrLivArea (> 4000 sqft)
outliers = df_clean[df_clean["GrLivArea"] > 4000][
    ["Id", "GrLivArea", "OverallQual", "SalePrice"]
]
print("=== Outlier Ekstrem pada GrLivArea (> 4000 sqft) ===")
print(outliers)

# 3. Penanganan Outlier: Hapus baris outlier ekstrem tersebut
print(f"\nJumlah baris sebelum penanganan outlier: {len(df_clean)}")
df_clean = df_clean[df_clean["GrLivArea"] <= 4000].copy()
print(f"Jumlah baris setelah penanganan outlier: {len(df_clean)}")
```

## **Transformasi Data (Encoding & Normalisasi/Standarisasi)**

```python
from sklearn.preprocessing import StandardScaler

# 1. Sesuaikan tipe data MSSubClass menjadi string/kategorikal
df_clean["MSSubClass"] = df_clean["MSSubClass"].astype(str)

# 2. Pisahkan Fitur (X) dan Target (y) serta Id
X_raw = df_clean.drop(columns=["Id", "SalePrice"])
y = df_clean["SalePrice"]

# 3. Lakukan One-Hot Encoding pada seluruh atribut kategorikal
X_encoded = pd.get_dummies(X_raw, drop_first=True)

# 4. Standarisasi seluruh fitur menggunakan StandardScaler
scaler = StandardScaler()
X_scaled = pd.DataFrame(
    scaler.fit_transform(X_encoded),
    columns=X_encoded.columns,
    index=X_encoded.index,
)

print("=== Hasil Transformasi Data ===")
print(f"Dimensi data fitur sebelum encoding  : {X_raw.shape}")
print(f"Dimensi data fitur setelah encoding : {X_encoded.shape}")
print(f"Dimensi data fitur setelah scaling  : {X_scaled.shape}")
print(
    f"\nRata-rata skala (sekitar 0) : {X_scaled.iloc[:, 0].mean():.4f}"
)
print(
    f"Standar deviasi skala (harus 1): {X_scaled.iloc[:, 0].std():.4f}"
)
```

## **Analisis Korelasi (Pembersihan Multikolinearitas)**

```python
# Hitung matriks korelasi absolut
corr_matrix = X_scaled.corr().abs()

# Ambil bagian segitiga atas dari matriks korelasi
upper_tri = corr_matrix.where(
    np.triu(np.ones(corr_matrix.shape), k=1).astype(bool)
)

# Cari fitur yang memiliki korelasi > 0.85 dengan fitur lain
to_drop = [
    column
    for column in upper_tri.columns
    if any(upper_tri[column] > 0.85)
]

print("=== Analisis Korelasi ===")
print(f"Jumlah fitur yang sangat berkorelasi (> 0.85): {len(to_drop)}")
print(f"Fitur yang dibuang: {to_drop}")

# Buang fitur berkorelasi tinggi
X_reduced = X_scaled.drop(columns=to_drop)
print(f"Dimensi data setelah reduksi korelasi: {X_reduced.shape}")
```

## P**CA**

```python
from sklearn.decomposition import PCA

# Inisialisasi PCA tanpa membatasi komponen untuk melihat total variance
pca_full = PCA()
pca_full.fit(X_reduced)

# Hitung kumulatif varians yang dijelaskan
cumulative_variance = np.cumsum(pca_full.explained_variance_ratio_)

# Cari jumlah komponen yang mempertahankan minimal 90% varians
n_components_90 = np.argmax(cumulative_variance >= 0.90) + 1

print("\n=== Hasil PCA ===")
print(
    f"Jumlah Komponen Utama yang dibutuhkan untuk 90% Varians: {n_components_90}"
)

# Visualisasi Kumulatif Variance
plt.figure(figsize=(10, 5))
plt.plot(
    range(1, len(cumulative_variance) + 1),
    cumulative_variance,
    marker="o",
    linestyle="--",
    color="b",
)
plt.axhline(y=0.90, color="r", linestyle="-", label="Target 90% Variance")
plt.axvline(
    x=n_components_90,
    color="green",
    linestyle=":",
    label=f"{n_components_90} Components",
)
plt.title("PCA Cumulative Explained Variance Ratio")
plt.xlabel("Number of Principal Components")
plt.ylabel("Cumulative Explained Variance")
plt.legend()
plt.tight_layout()
plt.show()

# Transformasi data menggunakan jumlah komponen optimal
pca_optimal = PCA(n_components=n_components_90)
X_pca = pca_optimal.fit_transform(X_reduced)

print(f"Dimensi data akhir setelah PCA: {X_pca.shape}")
```
